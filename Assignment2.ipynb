What is a dictionary in Python? List out all properties of Dictionary.

A dictionary in Python is a built-in data type used to store data in key-value pairs. Each key is unique and is used to access its corresponding value. Dictionaries are created using curly braces {} or the dict() constructor.  * Stores data as key-value pairs  * Keys are unique * Mutable *Values can be duplicated * Keys must be immutable (hashable) * Unordered *Dynamic size * Can store different data types *Indexed by keys, not positions *Supports nesting

How can you update dictionary values in Python? Provide examples. 

You can update dictionary values in Python in several ways.Update a single value using its key.Update multiple values using update().Add a new key-value pair using assignment.Update values using a loop.Using dictionary comprehension

```python
student = {
    "name": "John",
    "age": 25
}

student["age"] = 26

print(student)
```

```python
student = {
    "name": "John",
    "age": 25,
    "course": "Python"
}

student.update({
    "age": 26,
    "course": "C#"
})

print(student)
```

```python
student = {
    "name": "John",
    "age": 25
}

student["city"] = "Kochi"

print(student)
```

```python
marks = {
    "John": 70,
    "Alice": 80,
    "Bob": 75
}

for name in marks:
    marks[name] += 5

print(marks)
```

```python
marks = {
    "John": 70,
    "Alice": 80,
    "Bob": 75
}

new_marks = {name: mark + 5 for name, mark in marks.items()}

print(new_marks)
```

Count the frequency of each word in a given sentence using a dictionary. 

```python
sentence = "python is easy and python is powerful"
words = sentence.split()
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
```

Count the frequency of each character in a given word using a dictionary. 

```python
word = "malayalam"
frequency = {}
for char in word:
    frequency[char] = frequency.get(char, 0) + 1
print(frequency)
```

Sort a dictionary by keys and values.

```python
data = {"banana": 3, "apple": 5, "orange": 2}

sortKey = sorted(data)
print(sortKey)
sortValue = sorted(data.values())
print(sortValue)
sortKeyValue = sorted(data.items(), key=lambda x: x[1])
print(sortKeyValue)
```

Find the sum of all values in a dictionary. 

```python
data  = {"banana": 3, "apple": 5, "orange": 2}
result = sum(data.values())
print(result)
```

Create a dictionary using dictionary comprehension that contains even numbers as keys and
their squares as values from 1 to 10.

```python
result = {x: x ** 2 for x in range(1, 11) if x % 2 == 0}
print(result)
```
