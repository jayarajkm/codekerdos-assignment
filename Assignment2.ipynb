What is a dictionary in Python? List out all properties of Dictionary.

A dictionary in Python is a built-in data type used to store data in key-value pairs. Each key is unique and is used to access its corresponding value. Dictionaries are created using curly braces {} or the dict() constructor.  * Stores data as key-value pairs  * Keys are unique * Mutable *Values can be duplicated * Keys must be immutable (hashable) * Unordered *Dynamic size * Can store different data types *Indexed by keys, not positions *Supports nesting

Count the frequency of each word in a given sentence using a dictionary. 

```python
sentence = "python is easy and python is powerful"
words = sentence.split()
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
```

Count the frequency of each character in a given word using a dictionary. 

```python
word = "malayalam"
frequency = {}
for char in word:
    frequency[char] = frequency.get(char, 0) + 1
print(frequency)
```

. Group a list of words into lists of anagrams using a dictionary. Print the list of lists where each
inner list contains words that are anagrams of each other.
● Input: ["listen", "silent", "enlist", "hello", "lemon", "melon", "debitcard", "badcredit"]
● Anagram groups: [['listen', 'silent', 'enlist'], ['hello'], ['lemon', 'melon'], ['debitcard', 'badcredit']] 


```python
words = ["listen", "silent", "enlist", "hello", "lemon", "melon", "debitcard", "badcredit"]
anargrams = {}

for word in words:
    key = ''.join(sorted(word))
    if key not in anargrams:
        anargrams[key] = []
        
    anargrams[key].append(word)

result = list(anargrams.values())
print(result)
```

Find mirror characters in a string using a dictionary.
For example, 'A' mirrors to 'Z', 'B' mirrors to 'Y',
Input string: "Hello World"

```python
mirror_alpha = {}

for i in range(26):
    letter = chr(ord('a') + i)
    mirror_letter = chr(ord('z') - i)
    mirror_alpha[letter] = mirror_letter

for i in range(26):
    letter = chr(ord('A') + i)
    mirror_letter = chr(ord('Z') - i)
    mirror_alpha[letter] = mirror_letter

string = "Hello World"
result = ""

for char in string:
    if char in mirror_alpha:
        result += mirror_alpha[char]
    else:
        result += char

print(result)

    
```

Count the number of vowels in a given string and store the counts in a dictionary

Create a nested dictionary where the keys are letters from 'A' to 'E' and the values are
dictionaries mapping numbers from 1 to 5 to their squares.
Output = { 'A': {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}, 'B': {1: 1, 2: 4, 3: 9, 4: 16, 5: 25},...}

```python
result = {}

for i in range(5):
    square_root = {}
    letter = chr(ord('A')+ i)
    for j in range(1,6):
        square_root[j] = j**2
    result[letter] = square_root
print(result)
    
```

Explain the difference between a set and a frozenset. 

A set is mutable, meaning you can add or remove elements after creating it.A frozenset is an immutable set. Once created, you cannot add or remove elements.

Can you have a set of sets in Python? Explain with reasons. 

No, you cannot directly have a normal set containing other set objects.The reason is that elements of a set must be hashable, while a normal set is mutable and therefore unhashable.A set requires its elements to be hashable so Python can efficiently determine whether an element already exists.

Find the minimum and maximum elements in a set without using the built-in min() and max()
functions. 


```python
test = {45, 12, 78, 3, 56, 23, 89}
min = next(iter(test))
max = next(iter(test))
for i in test:
    if(i < min):
        min = i
    if(i > max):
        max = i
print("Minimum:", min)
print("Maximum:", max)
```
