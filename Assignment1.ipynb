What is a List? 

A list is collection of elements that are stored in a variable. List can be like arrays. It is Ordered and Mutable and can contain items of different data types. Also, it can store Duplicate Items. It is Contained in [] brackets. List can add, Remove and delete items.

What is a Tuple?

A tuple is a collection of items which are like a list, but it is immutable.  It is also ordered collection and can store duplicate values as well as different datatype items. It is contained in () brackets. Tuples cannot add, remove or updated items.

What is the difference between List and Tuple?

List is a collection of items which can be mutable, but Tuples are immutable. Lists have syntax as [] whereas tuples have () brackets. Tuple is Faster than lists.

Python Program to find the largest element in the list.

```python
a = [2,3,4,5,8,1,0]
largest = a[0]
for i in range(1, len(a)):
    if(a[i] > largest):
        largest = a[i]
print("Largest element is:", largest)
```

Given a list containing elements of different types (integers, strings, and floats), create three separate lists to store integers, strings, and floats respectively.

```python
a = [1,'abc',3.12,True, 5, 'xyz', 7.89, False]
integer_list = []
float_list = []
string_list = []
bool_list = []
for i in range(len(a)):
    if type(a[i]) == int:
        integer_list.append(a[i])
    elif type(a[i]) == float:
        float_list.append(a[i])
    elif type(a[i]) == str:
        string_list.append(a[i])
    elif type(a[i]) == bool:
        bool_list.append(a[i]) 
print("List of integers is:", integer_list)
print("List of floats is:", float_list)
print("List of strings is:", string_list)
print("List of booleans is:", bool_list)
```

Write a Python program to check whether a list contains a sublist

```python
a = [1,2,3,4,5,6,7,8,9]
b = [3,5]
found = False
for i in range(len(a) - (len(b) + 1)):
    if a[i:i+len(b)] == b:
        found = True
        break
print("Sublist found" if found else "sublist not found")
```

How to sort a list in ascending and descending order without using the sort function?

```python
a = [32,3412,213,123,43,12,324,546]

for i in range(len(a)):
    smallest = i
    for j in range(i+1, len(a)):
        if a[j] < a[smallest]:
            smallest = j
    a[i],a[smallest] = a[smallest],a[i]
print("Sorted list is:", a)
```

Given a list of integers, create a new list where each element is replaced by the next largest number in the original list. If there is no larger number, replace the element with -1. ● input_list = [4, 2, 6, 8, 1, 2] ● output_list = [6, 6, 8, -1, 2, -1]

```python
input_list = [4, 2, 6, 8, 1, 2] 
for i in range(len(input_list)):
    found = False
    for j in range(i+1,len(input_list)):
        if (input_list[i] < input_list[j]):
            found = True
            input_list[i] = input_list[j]
            break

    if not found:
        found = False
        input_list[i] = -1
print(input_list)
```

Write a function to reverse the elements within each tuple in a list of tuples. ● tuples = [(1, 2), (3, 4), (5, 6)] ● Output: [(2, 1), (4, 3), (6, 5)]

```python
input_list = [(1, 2), (3, 4), (5, 6)] 
new_list = []

for i in  range(len(input_list)):
    a = tuple(reversed(input_list[i]))
    new_list.append(a)

print("Original list of tuples:", input_list)
print("List of tuples after reversing:", new_list)
```
